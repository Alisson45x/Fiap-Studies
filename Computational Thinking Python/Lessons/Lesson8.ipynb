```python
#Ex.1 
def soma_ate(n):
    total = 0
    for i in range(1, n + 1):
        total = total + i
    return total
soma_ate(4)
```

```python
#EX 2
def som_ate_n(n):
    return n * (n + 1) // 2
som_ate_n(4)
```
