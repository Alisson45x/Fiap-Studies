```python
def fatorial_rec(n):
    if n == 0:
        return 1
    return n * fatorial_rec(n - 1)
fatorial_rec(5)
    
```

```python
number = int(input("Digite qualquer bosta ai: "))
def fatorial_lit(n):
    resultado = 1
    count = n
    print(f"!{n}", end="")
    for i in range(1, n+1):
        resultado *= i
        count -= 1
        print(f".{count}", end="")
        
    return print(f" = {resultado}")
fatorial_lit(number)
```
