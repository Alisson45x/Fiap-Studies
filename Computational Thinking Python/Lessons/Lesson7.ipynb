```python
fila = []
while True:
    cliente = input("Digite o nome do cliente e fim para acabar:")
    if cliente.lower() == "fim":
        break
    fila.append(cliente)
    print(f"O cliente: {cliente} foi adiconado na fila")
print("Todos os cliente da fila:")
for i in range(len(fila)):
    print(f"Nome: {fila[i]}")
while True:
    print(f"O cliente {fila[0]}, foi atendido")
    fila.pop(0)
    if len(fila) < 1:
        break
    print(f"cliente que ainda faltam:")
```
