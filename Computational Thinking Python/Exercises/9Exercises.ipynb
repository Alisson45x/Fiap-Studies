```python
class Aluno():
    def __init__(self, nome, n1, n2):
        self.nome = nome
        self.n1 = n1
        self.n2 = n2
        
    def calc_media(self):
        return (self.n1 + self.n2) / 2
    
    def situacao(self):
        if calc_media >= 7:
            return print("Aprovado")
        else:
            return print("Reprovado")
    
alisson = Aluno("Alisson", 9, 10)
print(alisson.calc_media())        
print(alisson.situacao())        
alisson.n1 = 2
print(alisson.situacao())        
                
```
