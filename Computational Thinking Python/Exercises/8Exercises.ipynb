```python
#Ex1
aluno = {}
aluno["Nome"] = input("Digite seu nome: ")
aluno["Nota1"] = int(input("Digite sua 1 nota: "))
aluno["Bruno"["n1"]] in input('DIgite a nota?')
aluno["Nota2"] = int(input("Digite sua 2 nota: "))
aluno["Nota3"] = int(input("Digite sua 3 nota: "))

media = (aluno["Nota1"] + aluno["Nota2"] + aluno["Nota3"]) / 3
print(f"{media:.2f}")
if media < 6:
    print(f"{aluno['Nome']} Reprovado")
else:
    print(f"{aluno['Nome']} Aprovado")
```

```python
#Ex2
conta = {
    "Titular": "Alisson",
    "Saldo": 1500
}
saque = int(input("Quanto quer sacar? "))
if saque > conta["Saldo"]:
    print("Saldo insuficiente!")
else:
    conta["Saldo"] = conta["Saldo"] - saque
    print(f"Restou {conta['Saldo']}")
```

```python
#Ex3
livro = {
    "total_paginas": 500,
    "paginas_lidas": 700
}
porcentagem = livro["paginas_lidas"] / livro["total_paginas"] * 100
paginas_faltantes = livro['total_paginas'] - livro['paginas_lidas']
print(f'{porcentagem:.0f}%')
if porcentagem >= 100:
    print("Todo livro foi lido")
else:
    print(f"Falta {paginas_faltantes} paginas para acabar o livro que é equivalente à {paginas_faltantes / livro['total_paginas'] * 100:.0f}%")
```
