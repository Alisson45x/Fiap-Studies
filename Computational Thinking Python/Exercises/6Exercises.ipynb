```python
#Ex 1


def fahren(a):
    fahrenheit = (a* 1.8) + 32
    return fahrenheit

temp = float(input("Digite uma temperatura em celsius: "))
print(f"A temperatura em celsius {temp} vira {fahren(temp)} em fahrenheit")
```

```python
#Ex 2
def escolha(figura):

    figura = int(input("Fale uma figura geometria entre: \n1.Circulo\n2.Triangulo\n3.Quadrado"))

    match figura:
        case 1:
            print("Area = π x r²")
            print("Circuferencia = 2 x π x r")
        case 2:
            print("Area = (b x h) / 2")
            print("Perimetro = a + b + c")
        case 3:
            print("Area = l x l")
            print("Perimetro = l x 4")
        case _:
            print("Figura invalida")
escolha(figura)
```
